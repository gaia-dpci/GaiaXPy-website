## XP data products

- `calibrate`, `convert` and `generate` accept DR3 continuous XP mean spectra.

<a id="gaiaxpy-local-files-and-pandas-objects" name="gaiaxpy-local-files-and-pandas-objects"></a>

## Local files and pandas objects

For mean spectra, supported local formats include CSV, ECSV, FITS, XML/VOTable, with supported gzip files and Archive ZIP downloads as provided by the Gaia Archive. A DataFrame can be passed directly. A Series represents one spectrum row.

```python
from gaiaxpy import calibrate

calibrated, sampling = calibrate("path/to/xp_mean_spectra.ecsv", save_file=False)
# If spectra is an existing DataFrame of continuous XP mean spectra:
calibrated, sampling = calibrate(spectra, save_file=False)
# To pass only the first row (a pandas Series):
calibrated, sampling = calibrate(spectra.iloc[0], save_file=False)
```

<a id="gaiaxpy-archive-retrieval" name="gaiaxpy-archive-retrieval"></a>

## Archive retrieval

Mean-spectrum Archive inputs accept tuples and lists of source IDs, or an ADQL query that returns `source_id`. For tuples and lists the source IDs can be integers or integer strings.

```python
from gaiaxpy import convert

source_ids = [5853498713190525696, "5762406957886626816"]
sampled, sampling_pwl = convert(source_ids, save_file=False)
```

## Required columns

The data obtained from the Gaia Archive contains all of the required columns. However, sometimes a user may want to create their own custom data to provide to the functions.

In this case, `get_required_columns` is useful. This function lists the required fields, alternatives and conditional requirements for a function. Additionally, important information is also included.

The function itself or just the function name can be passed as a parameter.

````python
from gaiaxpy import get_required_columns

requirements = get_required_columns("calibrate")
print(requirements)

# Optionally pass the function instead of its name:
#from gaiaxpy import calibrate, get_required_columns
#requirements = get_required_columns(calibrate)
````

The list of columns can be easily obtained from the `requirements` object by calling the different subsections:

````python
# The attributes of the object can be inspected with requirements.__dir__()

print(f"Calibrate required columns: {requirements.required}")
print()
print(f"Calibrate conditional columns: {requirements.conditional}")
````
