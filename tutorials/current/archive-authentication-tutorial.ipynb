# Access the Gaia Archive with GaiaXPy

GaiaXPy can retrieve Gaia DR3 XP mean spectra directly from the Gaia Archive if an ADQL query or a selection of source IDs (a list or tuple of integers or integer strings) if used. This tutorial uses `calibrate` throughout. The same Archive authentication options are available in other functions that retrieve spectra.

## Public access and request limits

**Signing in is recommended but not mandatory** for publicly available DR3 spectra. Following restrictions introduced on the Gaia Archive on **14 October 2026**, users making repeated requests are encouraged to [register for a Cosmos account](https://www.cosmos.esa.int/web/gaia-users/register) and authenticate. The Archive's response time may vary depending on whether users are authenticated.

An **HTTP 429 (Too Many Requests)** response indicates that the maximum number of requests permitted within a certain period has been exceeded. Reduce the request rate and retry later.

A public query can still be run without credentials:


````python
from gaiaxpy import calibrate

query = """
SELECT TOP 2 source_id
FROM gaiadr3.gaia_source
WHERE has_xp_continuous = 'True'
"""

calibrated_spectra, sampling = calibrate(query, save_file=False)
````

## Sign in with a username

For interactive use, pass `username` alone. GaiaXPy prompts you to enter your Cosmos password without displaying it. Even though it is allowed to match the underlying Astroquery implementation, **avoid passing your password directly as a function argument**, especially in notebooks or scripts that may be saved or shared.

The prompt requires an interactive session that supports hidden password entry.

````python
calibrated_spectra, sampling = calibrate(query, username="your_cosmos_username", save_file=False)
````

## Use a credentials file

For unattended processing, use `credentials_file` instead of `username`. Store this private UTF-8 file **outside the repository**, do not commit it to version control, and restrict access to your user account. Its contents must be exactly a username on the first line and a password on the second, with no labels or quotation marks. For example:

```text
example_cosmos_username
example_password
```

Then supply the path to the file. Do not combine `credentials_file` with `username` or `password`.


````python
calibrated_spectra, sampling = calibrate(query, credentials_file="/path/to/private/credentials.txt", save_file=False)
````

## Source-ID selections

A list or tuple of source IDs also works with `calibrate`. Authentication is optional for public data; pass `username` or `credentials_file` as above is recommended.

````python
source_ids = [5853498713190525696, 5762406957886626816]
calibrated_spectra, sampling = calibrate(source_ids, save_file=False)
````
